## 4. Prediction Task

The project treats 30-day readmission prediction as a **binary classification problem**.

The original `readmitted` variable is transformed as follows:

| Original Value | Meaning | Binary Label |
|:---|:---|:---:|
| `<30` | Readmitted within 30 days | `1` |
| `>30` | Readmitted after 30 days | `0` |
| `NO` | No recorded readmission | `0` |

**Positive class:** Readmission within 30 days  
**Negative class:** No recorded readmission within 30 days

The model produces:

1. A predicted probability of 30-day readmission.
2. A binary classification based on a selected decision threshold.

> **Prediction timing:** Predictions are assumed to be generated at discharge. Features must be restricted to information available by that point.

## 6. Evaluation Strategy

### Why Accuracy Alone Is Insufficient

The target classes may be imbalanced. A model that predicts the majority class for every encounter can achieve high accuracy while failing to identify patients who are readmitted within 30 days.

For this reason, an accuracy target such as **85%** is not sufficient evidence of a useful model.

### Evaluation Metrics

| Metric | Purpose |
|:---|:---|
| **Average Precision (AP)** | Summarizes precision–recall performance, with emphasis on the positive class. |
| **ROC-AUC** | Measures discrimination across classification thresholds. |
| **Recall / Sensitivity** | Measures the proportion of actual readmissions correctly identified. |
| **Precision** | Measures the proportion of predicted readmissions that are correct. |
| **F1-Score** | Balances precision and recall at a selected threshold. |
| **Balanced Accuracy** | Gives equal importance to sensitivity and specificity. |
| **Accuracy** | Reports overall correctness as a supplementary metric. |
| **Brier Score** | Assesses the quality of predicted probabilities. |
| **Confusion Matrix** | Displays true positives, true negatives, false positives, and false negatives. |

### Validation Principles

- Keep encounters from the same patient within a single data partition.
- Prevent information from the test set from influencing preprocessing or tuning.
- Apply resampling only within training folds.
- Compare all models using the same evaluation partitions.
- Select thresholds before evaluating the final test set.
- Report results transparently, including unsuccessful approaches.

```python
#Rebuild cell
import time
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
from sklearn.base import clone
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import roc_auc_score, brier_score_loss

df = pd.read_csv("../data/diabetic_data.csv")
df = df.replace("?", np.nan)
df["target"] = (df["readmitted"] == "<30").astype(int)
df = df[~df["discharge_disposition_id"].isin([11, 13, 14, 19, 20, 21])]
df = df.drop(columns=["encounter_id", "weight", "examide", "citoglipton"])
df["max_glu_serum"] = df["max_glu_serum"].fillna("NotTested")
df["A1Cresult"] = df["A1Cresult"].fillna("NotTested")
for col in ["race", "payer_code", "medical_specialty", "diag_1", "diag_2", "diag_3"]:
    df[col] = df[col].fillna("Unknown")

def icd_group(code):
    if code == "Unknown":
        return "Unknown"
    if code.startswith(("V", "E")):
        return "Other"
    n = int(float(code))
    if n == 250:
        return "Diabetes"
    if 390 <= n <= 459 or n == 785:
        return "Circulatory"
    if 460 <= n <= 519 or n == 786:
        return "Respiratory"
    if 520 <= n <= 579 or n == 787:
        return "Digestive"
    if 580 <= n <= 629 or n == 788:
        return "Genitourinary"
    if 710 <= n <= 739:
        return "Musculoskeletal"
    if 800 <= n <= 999:
        return "Injury"
    if 140 <= n <= 239:
        return "Neoplasms"
    return "Other"

for col in ["diag_1", "diag_2", "diag_3"]:
    df[col + "_group"] = df[col].apply(icd_group)
df = df.drop(columns=["diag_1", "diag_2", "diag_3"])

y = df["target"]
groups = df["patient_nbr"]
X = df.drop(columns=["target", "readmitted", "patient_nbr"])
id_cols = ["admission_type_id", "discharge_disposition_id", "admission_source_id"]
X[id_cols] = X[id_cols].astype(str)
numeric_cols = ["time_in_hospital", "num_lab_procedures", "num_procedures",
                "num_medications", "number_outpatient", "number_emergency",
                "number_inpatient", "number_diagnoses"]
categorical_cols = [c for c in X.columns if c not in numeric_cols]

preprocess = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
])
preprocess_gb = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_cols),
])
model = Pipeline([("prep", preprocess), ("clf", LogisticRegression(max_iter=1000))])
model_gb = Pipeline([("prep", preprocess_gb), ("clf", HistGradientBoostingClassifier(random_state=0))])
model_rf = Pipeline([("prep", preprocess_gb), ("clf", RandomForestClassifier(
    n_estimators=200, min_samples_leaf=5, n_jobs=-1, random_state=0))])
models = {"LogReg": model, "GradBoost": model_gb, "RandomForest": model_rf}

print(X.shape, y.shape, groups.nunique(), len(numeric_cols), len(categorical_cols))
```

Step 14: Build one 5-fold split and check it

What we’re doing: creating the splitting tool the professor asked for and testing that it behaves correctly. This is the center of my experimental design, so it is worth understanding well.

New words:

- 5-fold cross-validation: the data is cut into 5 parts (folds). The model trains on 4 parts and is tested on the remaining 1, and this is repeated 5 times so each part is the test set once. You get 5 scores for each model.
- Stratified: each fold keeps the same share of readmitted patients (about 11.4%).
- Grouped: all rows of one patient end up in the same fold, so a patient is never in both train and test.
- StratifiedGroupKFold is the scikit-learn tool that does all of this at once. It is a splitting tool only. It doesn’t train anything, and it is a small helper, but I’ll confirm with the professor that it is allowed.

My professor wants this 5-fold procedure repeated 20 times with different shuffles, which gives each model 100 scores. Today we only build one round and check it.

```python
from sklearn.model_selection import StratifiedGroupKFold

cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=0)

for i, (train_idx, test_idx) in enumerate(cv.split(X, y, groups)):
    overlap = len(set(groups.iloc[train_idx]) & set(groups.iloc[test_idx]))
    print(i, len(train_idx), len(test_idx),
          round(y.iloc[test_idx].mean(), 4), overlap)
    
#n_splits=5 means 5 folds. shuffle=True mixes the patients before splitting. random_state=0 fixes the shuffle, so you get the same result every time you run it. That is how we keep the experiment reproducible.
#cv.split(X, y, groups) produces 5 pairs of row positions: train_idx (the training rows) and test_idx (the test rows).
#overlap counts the patients who appear in both the training and the test part. It must be 0.
#y.iloc[test_idx].mean() is the share of readmissions in that test fold, because the mean of a 0/1 column is the share of ones.
```

My folds are correct. Every overlap is 0, so no patient appears in both train and test, and each test fold has the same share of readmissions (about 11.4%).

*Step 15: Train the first model on one fold*

What we’re doing: training logistic regression on the training part of fold 0, then measuring how well it predicts on the test part. This is my first real result, and it will be my baseline.

New words:

- One-hot encoding: a model can only work with numbers, so each category becomes its own 0/1 column. For example, gender becomes columns like gender_Male and gender_Female, with a 1 where the patient belongs.
- Scaling (StandardScaler): numeric columns are shifted and shrunk so they all have a similar size. Without it, a column with large values (like lab procedures) can dominate one with small values.
- Pipeline: a chain that does preprocessing and the model in a fixed order. This matters because the encoder and scaler learn from the data (for example, the average of each column). A pipeline makes sure they learn only from the training rows, never from the test rows, which would be another form of leakage.
- AUC: a score from 0.5 to 1. It says how often the model gives a higher risk to a patient who was readmitted than to one who was not. 0.5 means guessing, and 1.0 means perfect.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

train_idx, test_idx = next(iter(cv.split(X, y, groups)))

preprocess = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
])

model = Pipeline([
    ("prep", preprocess),
    ("clf", LogisticRegression(max_iter=1000)),
])

model.fit(X.iloc[train_idx], y.iloc[train_idx])
probs = model.predict_proba(X.iloc[test_idx])[:, 1]
print("AUC:", round(roc_auc_score(y.iloc[test_idx], probs), 4))

#next(iter(cv.split(...))) takes only the first pair of train and test rows (fold 0).
#ColumnTransformer applies the scaler to the numeric columns and the encoder to the categorical ones. handle_unknown="ignore" prevents an error if the test part contains a category the training part never saw.
#LogisticRegression(max_iter=1000) is the model. max_iter=1000 gives it enough steps to finish learning.
#model.fit(...) trains on the training rows only.
#predict_proba(...)[:, 1] gives, for each test patient, the probability of readmission. [:, 1] selects the column for class 1.
#roc_auc_score compares those probabilities with the true answers.
```

Step 16: Train the second model, gradient boosting

What we’re doing: training my black-box model on the same fold, so we can compare it fairly with logistic regression.

New word: gradient boosting builds many small decision trees one after another, and each new tree tries to fix the mistakes of the previous ones. It can capture more complex patterns than logistic regression, but you can’t read its decision directly from coefficients. That is why it is the “black box” in my project. We use scikit-learn’s HistGradientBoostingClassifier, which is a fast version of it.

Two things to keep the comparison fair:

- It uses the same train and test rows (train_idx and test_idx) and the same preprocessing steps as before.
- This version needs the one-hot columns as a full table, not a compact sparse one, so we add sparse_output=False.

```python
from sklearn.ensemble import HistGradientBoostingClassifier

preprocess_gb = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_cols),
])

model_gb = Pipeline([
    ("prep", preprocess_gb),
    ("clf", HistGradientBoostingClassifier(random_state=0)),
])

model_gb.fit(X.iloc[train_idx], y.iloc[train_idx])
probs_gb = model_gb.predict_proba(X.iloc[test_idx])[:, 1]
print("AUC:", round(roc_auc_score(y.iloc[test_idx], probs_gb), 4))

#preprocess_gb is the same recipe as before, with the one small change in the encoder.
#HistGradientBoostingClassifier(random_state=0) is the model. random_state=0 makes the result repeatable.
#The rest is identical to Step 15, so any difference in AUC comes from the model and not from the setup.
```

Step 17: Train the third model, Random Forest

What we’re doing: 
adding the third model that professor asked for. ANOVA needs at least three groups, so with this model we will have logistic regression, gradient boosting and Random Forest.

New word: 
a random forest builds many decision trees at the same time. Each tree sees a random part of the data and a random part of the columns, and the final prediction is the average of all the trees. It is a second black box, but a different kind from boosting, because its trees are independent instead of correcting each other.

```python
from sklearn.ensemble import RandomForestClassifier

model_rf = Pipeline([
    ("prep", preprocess_gb),
    ("clf", RandomForestClassifier(
        n_estimators=200, min_samples_leaf=5, n_jobs=-1, random_state=0
    )),
])

model_rf.fit(X.iloc[train_idx], y.iloc[train_idx])
probs_rf = model_rf.predict_proba(X.iloc[test_idx])[:, 1]
print("AUC:", round(roc_auc_score(y.iloc[test_idx], probs_rf), 4))

#We reuse preprocess_gb from the last step, because the forest also needs the full one-hot table. If you run this in a fresh session, run the previous cell first.
#n_estimators=200 means 200 trees.
#min_samples_leaf=5 means each tree’s final groups must contain at least 5 patients. This stops the trees from memorizing single patients, so the forest doesn’t overfit.
#n_jobs=-1 uses all the cores of your processor, so it runs faster.
#random_state=0 makes the result repeatable.
```

Step 18: Run the full procedure once (1 repeat × 5 folds × 3 models)

What we’re doing: 
putting everything into a loop, so that every model is trained and tested on all 5 folds of a repeat. We will get 15 scores (5 per model), and we’ll save them. We run one repeat first to check that the code works and to measure how long it takes. The 20 repeats come in the next step.

New words:

Repeat: one full round of 5-fold cross-validation with its own shuffle. The shuffle is controlled by a seed (a number), so with random_state=repeat every repeat splits the patients differently, and the experiment can still be reproduced exactly.
clone(model) makes a fresh, untrained copy of a model. Without it, a model would carry things it learned in the previous fold into the next one, which would be a leak.

```python
import time
from sklearn.base import clone

models = {
    "LogReg": model,
    "GradBoost": model_gb,
    "RandomForest": model_rf,
}

n_repeats = 1
rows = []
start = time.time()

for repeat in range(n_repeats):
    cv_r = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=repeat)
    for fold, (tr_idx, te_idx) in enumerate(cv_r.split(X, y, groups)):
        for name, mdl in models.items():
            fitted = clone(mdl)
            fitted.fit(X.iloc[tr_idx], y.iloc[tr_idx])
            p = fitted.predict_proba(X.iloc[te_idx])[:, 1]
            auc = roc_auc_score(y.iloc[te_idx], p)
            rows.append({"repeat": repeat, "fold": fold, "model": name, "auc": auc})
        print("repeat", repeat, "fold", fold, "done")

results = pd.DataFrame(rows)
results.to_csv("results_test.csv", index=False)
print(len(results), "scores in", round(time.time() - start), "seconds")
print(results.groupby("model")["auc"].agg(["mean", "std"]).round(4))

#The three nested loops go over repeats, folds and models. For every combination we train on the training rows and compute the AUC on the test rows.
#Each result is stored as one row with the repeat, fold, model name and AUC. At the end, pd.DataFrame(rows) turns the list into a table.
#to_csv(...) saves the table as a file in your notebook’s folder, so the results survive even if the notebook closes.
#The last line groups the table by model and shows the mean and the standard deviation of the AUC. The standard deviation shows how much a model’s score moves from fold to fold, which is the experimental noise your professor wants to measure.
```

Step 21: Run a one-way ANOVA

What we’re doing: 
testing whether the three models differ in AUC at all. This is the first statistical test the professor listed.

How to think about it:

# ANOVA starts from the assumption that all three models have the same true average AUC (the null hypothesis).
# It compares how far apart the three group means are with how much the scores vary inside each group.
# The result is an F value (larger means more evidence of a difference) and a p-value. If p is below 0.05, we reject the assumption that all means are equal.
# ANOVA only says that at least one model differs. It doesn’t say which one. That comes in the next test, Tukey’s HSD.

```python
from scipy import stats

model_names = ["LogReg", "GradBoost", "RandomForest"]
auc_groups = [results.loc[results["model"] == m, "auc"] for m in model_names]

F, p = stats.f_oneway(*auc_groups)
print(f"F = {F:.2f}, p = {p:.3g}")

# results.loc[results["model"] == m, "auc"] picks the 100 AUC scores of one model.
# auc_groups collects those three groups of scores into a list.
# stats.f_oneway(*auc_groups) runs the ANOVA. The * hands the three groups to the function as three separate inputs.
# The last line prints F with 2 decimals and p in a compact form, such as 3.2e-30 for a very small number.
```

An F of 129.37 with p = 3.9e-41 is far below 0.05, so we reject the assumption that all three models have the same average AUC. At least one model differs from the others.

Two cautions for how you word this in the report:

# A p-value that tiny should not be read as “41 zeros of certainty”. Because the 100 scores per model overlap in their training data (the caution from last step), the true p-value is larger than this. The conclusion that the models differ is still safe, since the boxplot also shows it.
# ANOVA says nothing about which models differ. For that we need a post-hoc test.

Step 22: Tukey’s HSD, which pairs differ?

What we’re doing: 
comparing every pair of models: LogReg vs GradBoost, LogReg vs RandomForest, and GradBoost vs RandomForest.

Why not just run three separate tests? 
Each test has a 5% chance of a false alarm, and with several tests the chance that at least one of them is a false alarm grows. This is the family-wise error rate your professor mentioned. Tukey’s HSD (Honestly Significant Difference) adjusts for it, so the overall false-alarm rate across all pairs stays at 5%.

```python
tukey = stats.tukey_hsd(*auc_groups)
print(tukey)

ci = tukey.confidence_interval(confidence_level=0.95)
print(ci.low.round(4))
print(ci.high.round(4))

#stats.tukey_hsd(*auc_groups) runs the test on the same three groups as before.
#print(tukey) prints a table with one row per pair, showing the difference in means, the p-value and a 95% confidence interval.
#The two lines with ci print the lower and upper ends of those intervals as separate matrices.
```

All three pairs come out significant at the 5% level, but two of them are strong and one is borderline.

How to read the table (it lists each pair twice, once in each direction, so read only (0 - 1), (0 - 2) and (1 - 2)):

Pair	Mean difference in AUC	p-value	95% interval
LogReg vs GradBoost	LogReg lower by 0.011	about 0.000	-0.012 to -0.009
LogReg vs RandomForest	LogReg lower by 0.009	about 0.000	-0.011 to -0.007
GradBoost vs RandomForest	GradBoost higher by 0.002	0.037	0.000 to 0.003


# Both black boxes beat logistic regression. The intervals are far from 0, so this is a solid result.
# GradBoost vs RandomForest is borderline. The p-value of 0.037 is just under 0.05, and the lower end of the interval is almost exactly 0 (0.0001 in the matrix). So the evidence is weak.
# This borderline result needs a careful interpretation. As we saw, the scores overlap in their training data, so these p-values are somewhat too optimistic. If the correction were applied, 0.037 might well end up above 0.05. A fair wording would be: “a small difference of about 0.002 AUC, only marginally significant, and not robust to the dependence between folds.” Don’t write that GradBoost is better.
# Even if it is real, a difference of 0.002 AUC is practically tiny. Statistical significance and practical importance are two different questions, and your professor will like seeing that you separate them.

Step 23: Wilcoxon signed-rank test, GradBoost vs RandomForest

What we’re doing: 
testing the closest pair again with a second, different kind of test, to see whether it agrees with Tukey.

Why this test fits:
# It is non-parametric, which means it does not assume that the scores follow a bell-shaped curve, unlike ANOVA and Tukey.
# It is paired. For each of the 100 test folds, GradBoost and RandomForest were evaluated on exactly the same patients. The test compares the two scores fold by fold, which is more sensitive than comparing two separate groups.

```python
wide = results.pivot_table(index=["repeat", "fold"], columns="model", values="auc")
print(wide.shape)

diff = wide["GradBoost"] - wide["RandomForest"]
print("mean difference:", round(diff.mean(), 4))
print("GradBoost better in", (diff > 0).sum(), "of", len(diff), "folds")

stat, p_w = stats.wilcoxon(wide["GradBoost"], wide["RandomForest"])
print(f"Wilcoxon: statistic = {stat:.1f}, p = {p_w:.3g}")

#pivot_table(...) reshapes the results so that each row is one fold (a repeat and fold number) and each model has its own column. Then the two scores for the same fold sit side by side.
#diff is GradBoost minus RandomForest for every fold.
#The two print lines give the average gap and in how many of the 100 folds GradBoost scored higher.
#stats.wilcoxon(a, b) runs the paired test and gives a statistic and a p-value.
```

```python
import pandas as pd
import numpy as np
from scipy import stats

results = pd.read_csv("results.csv")
wide = results.pivot_table(index=["repeat", "fold"], columns="model", values="auc")
diff = wide["GradBoost"] - wide["RandomForest"]
print(wide.shape, round(diff.mean(), 4))
```

```python
J = len(diff)
ratio = 0.25

t_naive = diff.mean() / np.sqrt(diff.var(ddof=1) / J)
t_corr = diff.mean() / np.sqrt((1 / J + ratio) * diff.var(ddof=1))

p_naive = 2 * stats.t.sf(abs(t_naive), df=J - 1)
p_corr = 2 * stats.t.sf(abs(t_corr), df=J - 1)

print(f"naive:     t = {t_naive:.2f}, p = {p_naive:.3g}")
print(f"corrected: t = {t_corr:.2f}, p = {p_corr:.3g}")
```

```python
pairs = [("GradBoost", "LogReg"),
         ("RandomForest", "LogReg"),
         ("GradBoost", "RandomForest")]

for a, b in pairs:
    d = wide[a] - wide[b]
    t = d.mean() / np.sqrt((1 / J + 0.25) * d.var(ddof=1))
    p = 2 * stats.t.sf(abs(t), df=J - 1)
    print(f"{a} vs {b}: mean diff = {d.mean():.4f}, corrected t = {t:.2f}, p = {p:.3g}")
```

```python
rows = []
for m in ["LogReg", "GradBoost", "RandomForest"]:
    s = wide[m]
    mean = s.mean()
    sd = s.std(ddof=1)
    half = stats.t.ppf(0.975, df=J - 1) * np.sqrt((1 / J + 0.25) * s.var(ddof=1))
    rows.append({"model": m, "mean_auc": mean, "std": sd,
                 "ci_low": mean - half, "ci_high": mean + half})

summary = pd.DataFrame(rows).round(4)
summary.to_csv("summary_auc.csv", index=False)
print(summary)
```

```python
import time
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
from sklearn.base import clone
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import roc_auc_score, brier_score_loss

df = pd.read_csv("../data/diabetic_data.csv")
df = df.replace("?", np.nan)
df["target"] = (df["readmitted"] == "<30").astype(int)
df = df[~df["discharge_disposition_id"].isin([11, 13, 14, 19, 20, 21])]
df = df.drop(columns=["encounter_id", "weight", "examide", "citoglipton"])
df["max_glu_serum"] = df["max_glu_serum"].fillna("NotTested")
df["A1Cresult"] = df["A1Cresult"].fillna("NotTested")
for col in ["race", "payer_code", "medical_specialty", "diag_1", "diag_2", "diag_3"]:
    df[col] = df[col].fillna("Unknown")

def icd_group(code):
    if code == "Unknown":
        return "Unknown"
    if code.startswith(("V", "E")):
        return "Other"
    n = int(float(code))
    if n == 250:
        return "Diabetes"
    if 390 <= n <= 459 or n == 785:
        return "Circulatory"
    if 460 <= n <= 519 or n == 786:
        return "Respiratory"
    if 520 <= n <= 579 or n == 787:
        return "Digestive"
    if 580 <= n <= 629 or n == 788:
        return "Genitourinary"
    if 710 <= n <= 739:
        return "Musculoskeletal"
    if 800 <= n <= 999:
        return "Injury"
    if 140 <= n <= 239:
        return "Neoplasms"
    return "Other"

for col in ["diag_1", "diag_2", "diag_3"]:
    df[col + "_group"] = df[col].apply(icd_group)
df = df.drop(columns=["diag_1", "diag_2", "diag_3"])

y = df["target"]
groups = df["patient_nbr"]
X = df.drop(columns=["target", "readmitted", "patient_nbr"])
id_cols = ["admission_type_id", "discharge_disposition_id", "admission_source_id"]
X[id_cols] = X[id_cols].astype(str)
numeric_cols = ["time_in_hospital", "num_lab_procedures", "num_procedures",
                "num_medications", "number_outpatient", "number_emergency",
                "number_inpatient", "number_diagnoses"]
categorical_cols = [c for c in X.columns if c not in numeric_cols]

preprocess = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
])
preprocess_gb = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_cols),
])
model = Pipeline([("prep", preprocess), ("clf", LogisticRegression(max_iter=1000))])
model_gb = Pipeline([("prep", preprocess_gb), ("clf", HistGradientBoostingClassifier(random_state=0))])
model_rf = Pipeline([("prep", preprocess_gb), ("clf", RandomForestClassifier(
    n_estimators=200, min_samples_leaf=5, n_jobs=-1, random_state=0))])
models = {"LogReg": model, "GradBoost": model_gb, "RandomForest": model_rf}

print(X.shape, y.shape, groups.nunique(), len(numeric_cols), len(categorical_cols))
```

```python
rows = []
start = time.time()

for repeat in range(20):
    cv_r = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=repeat)
    for fold, (tr_idx, te_idx) in enumerate(cv_r.split(X, y, groups)):
        for name, mdl in models.items():
            fitted = clone(mdl)
            fitted.fit(X.iloc[tr_idx], y.iloc[tr_idx])
            p = fitted.predict_proba(X.iloc[te_idx])[:, 1]
            rows.append({
                "repeat": repeat, "fold": fold, "model": name,
                "auc": roc_auc_score(y.iloc[te_idx], p),
                "brier": brier_score_loss(y.iloc[te_idx], p),
            })
    pd.DataFrame(rows).to_csv("results_full.csv", index=False)
    print("repeat", repeat, "done,", round((time.time() - start) / 60, 1), "min elapsed")

results_full = pd.DataFrame(rows)
print(results_full.groupby("model")[["auc", "brier"]].mean().round(4))
```

```python
#Step 33: Rank the original features, not the one-hot columns
lr = clone(models["LogReg"])
lr.fit(X, y)

prep = lr.named_steps["prep"]
coefs = lr.named_steps["clf"].coef_[0]
ohe = prep.named_transformers_["cat"]

feature_of_column = numeric_cols + [
    col for col, cats in zip(categorical_cols, ohe.categories_) for _ in cats
]
print(len(feature_of_column))

Xs = X.sample(20000, random_state=0)
Xt = prep.transform(Xs).toarray()

contrib = (Xt - Xt.mean(axis=0)) * coefs
contrib_df = pd.DataFrame(contrib, columns=feature_of_column)
per_feature = contrib_df.T.groupby(level=0).sum().T

lr_importance = per_feature.abs().mean().sort_values(ascending=False)
print(lr_importance.head(10).round(4))
```

```python
#Step 34: Feature ranking for the black box with permutation importance
from sklearn.inspection import permutation_importance

cv0 = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=0)
tr_idx, te_idx = next(iter(cv0.split(X, y, groups)))

gb = clone(models["GradBoost"])
gb.fit(X.iloc[tr_idx], y.iloc[tr_idx])

X_te = X.iloc[te_idx].sample(8000, random_state=0)
y_te = y.loc[X_te.index]

r = permutation_importance(gb, X_te, y_te, scoring="roc_auc",
                           n_repeats=5, random_state=0, n_jobs=1)

gb_importance = pd.Series(r.importances_mean, index=X.columns).sort_values(ascending=False)
print(gb_importance.head(10).round(4))
```

```python
#Step 35: Compare the two rankings with Spearman correlation
from scipy import stats

lr_fold = clone(models["LogReg"])
lr_fold.fit(X.iloc[tr_idx], y.iloc[tr_idx])

r2 = permutation_importance(lr_fold, X_te, y_te, scoring="roc_auc",
                            n_repeats=5, random_state=0, n_jobs=1)
lr_perm = pd.Series(r2.importances_mean, index=X.columns).sort_values(ascending=False)
print(lr_perm.head(6).round(4))

rho_same, _ = stats.spearmanr(lr_perm.reindex(X.columns), gb_importance.reindex(X.columns))
rho_contrib, _ = stats.spearmanr(lr_importance.reindex(X.columns), gb_importance.reindex(X.columns))
print("Spearman, permutation vs permutation:", round(rho_same, 3))
print("Spearman, LR contribution vs GB permutation:", round(rho_contrib, 3))
```
